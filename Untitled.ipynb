# Word Segmentation attempt 1

Using MMC I did this. (MMC+ ZCR + STE)

```python
import sounddevice as sd
import numpy as np
import librosa
import librosa.display
import matplotlib.pyplot as plt
from sklearn.svm import SVC
from sklearn.preprocessing import StandardScaler
from IPython.display import Audio, display

# Step 1: Record Audio in Real-Time
def record_audio(duration=7, samplerate=16000):
    print("Recording... Speak now!")
    audio = sd.rec(int(duration * samplerate), samplerate=samplerate, channels=1, dtype="float32")
    sd.wait()  # Wait for the recording to finish
    print("Recording complete!")
    return audio.flatten(), samplerate

# Step 2: Ensure Equal Length for Features
def ensure_equal_length(zcr, ste):
    min_length = min(len(zcr), len(ste))
    zcr = zcr[:min_length]
    ste = ste[:min_length]
    return zcr, ste

# Step 3: MMC Clustering Algorithm
def mmc_clustering(features):
    # Standardize features for better SVM performance
    scaler = StandardScaler()
    features = scaler.fit_transform(features.T)

    # Initialize labels with k-means-like strategy (binary labels)
    initial_labels = np.random.choice([0, 1], size=features.shape[0])

    # Initialize SVM hyperplane parameters
    clf = SVC(kernel="linear", C=1.0)

    # Iterative optimization
    labels = initial_labels
    for _ in range(10):  # Limit the iterations for practicality
        # Train SVM with current labels
        clf.fit(features, labels)

        # Update labels based on the SVM decision function
        labels = (clf.decision_function(features) > 0).astype(int)
    return labels

# Step 4: Segment Audio
def segment_audio(y, sr, zcr, ste, hop_length, zcr_threshold=0.2, ste_threshold=0.01):
    # Ensure ZCR and STE have the same length
    zcr, ste = ensure_equal_length(zcr, ste)
    
    # Combine features for clustering
    features = np.vstack([zcr, ste])
    labels = mmc_clustering(features)

    # Identify boundaries based on clustering
    boundaries = []
    in_segment = False
    start_time = 0
    for i, label in enumerate(labels):
        if label == 1 and not in_segment:  # Word starts
            in_segment = True
            start_time = i * hop_length / sr
        elif label == 0 and in_segment:  # Word ends
            in_segment = False
            end_time = i * hop_length / sr
            boundaries.append((start_time, end_time))
    
    # Extract segments
    segments = []
    for start, end in boundaries:
        start_sample = int(start * sr)
        end_sample = int(end * sr)
        segments.append((start, end, y[start_sample:end_sample]))
    
    return segments, boundaries

# Step 5: Display Results and Play Segments
def display_segments(segments, sr, zcr, ste, boundaries, y):
    print("Detected Words:")
    for i, (start, end, segment) in enumerate(segments):
        print(f"Word {i+1}: Start = {start:.2f}s, End = {end:.2f}s, Duration = {end - start:.2f}s")
        display(Audio(segment, rate=sr))

    # Plot ZCR and STE with boundaries
    zcr, ste = ensure_equal_length(zcr, ste)
    time_axis = np.linspace(0, len(y) / sr, len(zcr))
    plt.figure(figsize=(12, 6))
    plt.plot(time_axis, zcr, label="ZCR")
    plt.plot(time_axis, ste, label="STE")
    plt.xlabel("Time (s)")
    plt.ylabel("Normalized Value")
    plt.title("ZCR and STE with Detected Boundaries")
    plt.legend()
    for start, end in boundaries:
        plt.axvspan(start, end, color="red", alpha=0.3)
    plt.show()

# Step 6: Main Execution
if __name__ == "__main__":
    # Record audio in real-time
    recorded_audio, sr = record_audio(duration=7)  # Adjust duration as needed

    # Compute ZCR and STE
    zcr = librosa.feature.zero_crossing_rate(recorded_audio, frame_length=2048, hop_length=512)[0]
    frames = librosa.util.frame(recorded_audio, frame_length=2048, hop_length=512)
    ste = np.sum(frames**2, axis=0)

    # Segment the recorded audio using MMC and refine with ZCR/STE
    segments, boundaries = segment_audio(recorded_audio, sr, zcr, ste, hop_length=512)

    # Display and play the segmented words
    display_segments(segments, sr, zcr, ste, boundaries, recorded_audio)
```

# Word Segmentation attempt 2

Without using MMC I did this. ( Just STE + ZCR)

```python
import sounddevice as sd
import numpy as np
import librosa
import librosa.display
import matplotlib.pyplot as plt
from IPython.display import Audio, display

# Step 1: Record Audio in Real-Time
def record_audio(duration=7, samplerate=16000):
    print("Recording... Speak now!")
    audio = sd.rec(int(duration * samplerate), samplerate=samplerate, channels=1, dtype="float32")
    sd.wait()  # Wait for the recording to finish
    print("Recording complete!")
    return audio.flatten(), samplerate

# Step 2: Segment Audio
def segment_audio(y, sr, frame_length=2048, hop_length=512, zcr_threshold=0.2, ste_threshold=0.01):
    # Compute features
    zcr = librosa.feature.zero_crossing_rate(y, frame_length=frame_length, hop_length=hop_length)[0]
    frames = librosa.util.frame(y, frame_length=frame_length, hop_length=hop_length)
    ste = np.sum(frames**2, axis=0)

    # Normalize features
    zcr_norm = zcr / np.max(zcr)
    ste_norm = ste / np.max(ste)

    # Detect word boundaries
    boundaries = []
    in_segment = False
    for i, (z, e) in enumerate(zip(zcr_norm, ste_norm)):
        if z > zcr_threshold or e > ste_threshold:
            if not in_segment:
                in_segment = True
                start_time = i * hop_length / sr
        else:
            if in_segment:
                in_segment = False
                end_time = i * hop_length / sr
                boundaries.append((start_time, end_time))
    
    # Extract segments
    segments = []
    for start, end in boundaries:
        start_sample = int(start * sr)
        end_sample = int(end * sr)
        segments.append((start, end, y[start_sample:end_sample]))
    return segments, zcr_norm, ste_norm, boundaries

# Step 3: Display Results and Play Segments
def display_segments(segments, sr, zcr_norm, ste_norm, boundaries, y):
    print("Detected Words:")
    for i, (start, end, segment) in enumerate(segments):
        print(f"Word {i+1}: Start = {start:.2f}s, End = {end:.2f}s, Duration = {end - start:.2f}s")
        display(Audio(segment, rate=sr))

    # Plot ZCR and STE
    plt.figure(figsize=(12, 6))
    plt.plot(np.linspace(0, len(y)/sr, len(zcr_norm)), zcr_norm, label="ZCR")
    plt.plot(np.linspace(0, len(y)/sr, len(ste_norm)), ste_norm, label="STE")
    plt.xlabel("Time (s)")
    plt.ylabel("Normalized Value")
    plt.title("ZCR and STE with Detected Boundaries")
    plt.legend()
    for start, end in boundaries:
        plt.axvspan(start, end, color="red", alpha=0.3)
    plt.show()

# Step 4: Main Execution
# Record audio in real-time
recorded_audio, sr = record_audio(duration=7)  # Adjust duration as needed

# Segment the recorded audio
segments, zcr_norm, ste_norm, boundaries = segment_audio(recorded_audio, sr)

# Display and play the segmented words
display_segments(segments, sr, zcr_norm, ste_norm, boundaries, recorded_audio)
```
